# 🌌 Qfóton: Interactive Quantum Photonics Quickstart

Run full-stack linear optical quantum computing simulations and Clements MZI matrix decompositions directly in your browser.

```python
# Clone repository and import core simulator engines
!git clone https://github.com/atharveeee-netizen/qfoton.git
%cd qfoton
import numpy as np
import matplotlib.pyplot as plt
from simulator.clements_compiler import ClementsCompiler
from simulator.sfwm_source import SFWMPhotonSource
from simulator.hom_interference import HongOuMandelSimulator
print('Qfóton engines imported successfully!')
```

## 1. On-Chip Single-Photon Pair Generation (SFWM)

```python
source = SFWMPhotonSource(radius_um=15.0, q_factor=100000.0)
res = source.simulate_pair_generation(pump_power_mw=5.0)
print(f'Coincidence-to-Accidental Ratio (CAR): {res["car_ratio"]}')
print(f'Heralded Single-Photon Purity g^(2)(0): {res["heralded_purity_g2"]}')
```

## 2. Universal Clements MZI Decomposition for 4x4 Unitary

```python
compiler = ClementsCompiler(num_modes=4)
U_rand = compiler.generate_random_unitary()
mzi_mesh = compiler.decompose(U_rand)
print(f'Total Mach-Zehnder Interferometers: {len(mzi_mesh["mzi_list"])}')
print(f'Reconstructed Unitary Fidelity: {mzi_mesh["decomposition_fidelity"] * 100:.4f}%')
```

## 3. Hong-Ou-Mandel Quantum Interference Dip

```python
hom = HongOuMandelSimulator(optical_bandwidth_nm=2.0)
hom_res = hom.scan_hom_dip(delay_range_ps=(-3.0, 3.0))
print(f'Calculated Quantum HOM Visibility: {hom_res["visibility"] * 100:.2f}%')
```
